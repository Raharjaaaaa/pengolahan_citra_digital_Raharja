## Tujuan
Menganalisis distribusi intensitas piksel pada citra grayscale menggunakan histogram, lalu membandingkan hasil manipulasi brightness dan contrast.

Foto yang digunakan adalah foto kegiatan di dalam ruangan yang disertakan pada notebook ini.

## 1. Import library dan memuat citra
Citra disimpan langsung di dalam notebook sehingga tidak perlu mengunggah file foto secara terpisah.

```python
import base64
from io import BytesIO
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Foto sumber yang disertakan di notebook
image_b64 = ''''''
image_bytes = base64.b64decode(image_b64)
img = Image.open(BytesIO(image_bytes)).convert('RGB')
gray = img.convert('L')
gray_np = np.array(gray)

print('Ukuran citra:', img.size)
print('Mode citra asli:', img.mode)
print('Ukuran citra grayscale:', gray_np.shape)

plt.figure(figsize=(7, 8))
plt.imshow(img)
plt.title('Citra Asli')
plt.axis('off')
plt.show()

plt.figure(figsize=(7, 5))
plt.imshow(gray_np, cmap='gray', vmin=0, vmax=255)
plt.title('Citra Grayscale')
plt.axis('off')
plt.show()
```

## 2. Histogram grayscale
Histogram menunjukkan jumlah piksel pada setiap tingkat intensitas dari 0 (hitam) sampai 255 (putih).

```python
plt.figure(figsize=(10, 4))
plt.hist(gray_np.ravel(), bins=256, range=(0, 256), color='gray')
plt.title('Histogram Citra Grayscale Asli')
plt.xlabel('Intensitas piksel (0–255)')
plt.ylabel('Jumlah piksel')
plt.xlim(0, 255)
plt.grid(alpha=0.2)
plt.show()

mean_intensity = float(gray_np.mean())
std_intensity = float(gray_np.std())
print(f'Rata-rata intensitas: {mean_intensity:.2f}')
print(f'Simpangan baku intensitas: {std_intensity:.2f}')
print('Intensitas rendah (0–85):', int(np.sum(gray_np < 86)))
print('Intensitas sedang (86–170):', int(np.sum((gray_np >= 86) & (gray_np <= 170))))
print('Intensitas tinggi (171–255):', int(np.sum(gray_np > 170)))
```

## 3. Manipulasi brightness
Brightness menggeser nilai intensitas piksel. Rumus yang digunakan adalah `hasil = piksel + beta`. Nilai dibatasi pada rentang 0–255 agar tetap valid.

```python
def change_brightness(image_array, beta):
    result = image_array.astype(np.int16) + beta
    return np.clip(result, 0, 255).astype(np.uint8)

brightness_minus40 = change_brightness(gray_np, -40)
brightness_plus40 = change_brightness(gray_np, 40)

brightness_results = [
    ('Brightness beta = -40', brightness_minus40),
    ('Brightness beta = +40', brightness_plus40)
]

fig, axes = plt.subplots(2, 2, figsize=(12, 9))
for ax, (title, arr) in zip(axes.ravel(), brightness_results):
    ax.imshow(arr, cmap='gray', vmin=0, vmax=255)
    ax.set_title(title)
    ax.axis('off')
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, (title, arr) in zip(axes, brightness_results):
    ax.hist(arr.ravel(), bins=256, range=(0, 256), color='gray')
    ax.set_title('Histogram ' + title)
    ax.set_xlabel('Intensitas piksel')
    ax.set_ylabel('Jumlah piksel')
    ax.set_xlim(0, 255)
plt.tight_layout()
plt.show()
```

### Analisis brightness
- **Beta = -40:** nilai intensitas dikurangi 40 sehingga citra menjadi lebih gelap. Histogram cenderung bergeser ke kiri; sebagian nilai yang melewati 0 akan terpotong menjadi 0.
- **Beta = +40:** nilai intensitas ditambah 40 sehingga citra menjadi lebih terang. Histogram cenderung bergeser ke kanan; sebagian nilai yang melewati 255 akan terpotong menjadi 255.

Perhatikan hasil grafik setelah kode dijalankan untuk melihat jumlah piksel yang mengalami pemotongan pada batas intensitas.

## 4. Manipulasi contrast
Contrast diubah dengan mengalikan intensitas terhadap faktor alpha: `hasil = alpha × piksel`. Alpha di bawah 1 mengurangi perbedaan intensitas, sedangkan alpha di atas 1 memperbesar perbedaan intensitas terhadap hitam.

```python
def change_contrast(image_array, alpha):
    result = image_array.astype(np.float32) * alpha
    return np.clip(result, 0, 255).astype(np.uint8)

contrast_05 = change_contrast(gray_np, 0.5)
contrast_18 = change_contrast(gray_np, 1.8)

contrast_results = [
    ('Contrast alpha = 0.5', contrast_05),
    ('Contrast alpha = 1.8', contrast_18)
]

fig, axes = plt.subplots(1, 2, figsize=(12, 7))
for ax, (title, arr) in zip(axes, contrast_results):
    ax.imshow(arr, cmap='gray', vmin=0, vmax=255)
    ax.set_title(title)
    ax.axis('off')
plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, (title, arr) in zip(axes, contrast_results):
    ax.hist(arr.ravel(), bins=256, range=(0, 256), color='gray')
    ax.set_title('Histogram ' + title)
    ax.set_xlabel('Intensitas piksel')
    ax.set_ylabel('Jumlah piksel')
    ax.set_xlim(0, 255)
plt.tight_layout()
plt.show()
```

### Analisis contrast
- **Alpha = 0.5:** seluruh intensitas diperkecil setengahnya. Citra cenderung lebih gelap dan rentang intensitas menjadi lebih sempit, sehingga perbedaan antara area terang dan gelap berkurang.
- **Alpha = 1.8:** intensitas diperbesar. Perbedaan area gelap dan terang lebih menonjol, tetapi piksel yang hasilnya melebihi 255 menjadi putih karena pemotongan nilai.

Efek akhir dapat diamati langsung pada citra dan histogram yang dihasilkan.

## 5. Perbandingan hasil dan kesimpulan
Jalankan sel berikut untuk membandingkan rata-rata intensitas dan simpangan baku citra asli dengan setiap hasil manipulasi.

```python
all_results = {
    'Asli grayscale': gray_np,
    'Brightness -40': brightness_minus40,
    'Brightness +40': brightness_plus40,
    'Contrast 0.5': contrast_05,
    'Contrast 1.8': contrast_18,
}

print(f"{'Citra':<22} {'Rata-rata':>12} {'Simpangan baku':>18} {'Piksel 0':>12} {'Piksel 255':>12}")
print('-' * 78)
for name, arr in all_results.items():
    print(f'{name:<22} {arr.mean():>12.2f} {arr.std():>18.2f} {np.sum(arr == 0):>12} {np.sum(arr == 255):>12}')

print('\nKESIMPULAN:')
print('- Brightness -40 membuat citra lebih gelap, sedangkan +40 membuatnya lebih terang.')
print('- Contrast 0.5 mengurangi perbedaan intensitas dan biasanya membuat citra tampak lebih datar.')
print('- Contrast 1.8 meningkatkan perbedaan intensitas, tetapi dapat menghilangkan detail pada area yang terlalu terang.')
print('- Untuk kebutuhan deteksi objek otomatis, tidak ada satu pengaturan yang selalu terbaik. Pilih hasil yang memperjelas objek target tanpa menghilangkan detail penting, lalu uji dengan metode deteksi yang akan digunakan.')
```
